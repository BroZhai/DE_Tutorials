```python
# 1. Check whether a Conda environment is available

import subprocess

def check_conda_environment():
    conda_exe = r"D:\Anaconda3\condabin\conda.bat"
    try:
        # 1. Check whether Conda exists and obtain its version
        conda_version = subprocess.check_output([conda_exe, "--version"]).decode("utf-8").strip()
        print(f"✅ Conda environment detected. Current version: {conda_version}\n")

        # 2. Obtain the names and installation paths of all Conda environments
        print("--- Existing Conda environments and paths ---")
        conda_envs = subprocess.check_output([conda_exe, "info", "--envs"]).decode("utf-8")
        # Filter out comment lines and print the remainder
        for line in conda_envs.split('\n'):
            if line and not line.startswith('#'):
                print(line)

    except FileNotFoundError:
        print("❌ No Conda environment was detected in this Colab instance.")
        print("💡 Tip: Colab uses pip by default. To use Conda, install and configure it with the `condacolab` package or a Miniconda script.")

# Run the check
check_conda_environment()
```
